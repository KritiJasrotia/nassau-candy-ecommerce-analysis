```python
#Region Analysis
region_analysis = df.groupby('Region').agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).sort_values('Sales', ascending=False)

region_analysis
```

```python
region_analysis['Profit Margin %'] = (
    region_analysis['Gross_Profit'] / region_analysis['Sales']
) * 100

region_analysis
```

```python
#Which Division generates the highest Sales and Gross Profit?
division_analysis = df.groupby('Division').agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).sort_values('Sales', ascending=False)

division_analysis
```

```python
division_analysis['Profit Margin %'] = (
    division_analysis['Gross_Profit'] /
    division_analysis['Sales']
) * 100

division_analysis
```

```python
#Product Analysis
product_analysis = df.groupby('Product Name').agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).sort_values('Sales', ascending=False)

product_analysis.head(10)
```

```python
product_analysis['Profit Margin %'] = (
    product_analysis['Gross_Profit'] /
    product_analysis['Sales']
) * 100
```

```python
product_analysis.head(10)
```

```python
top_profit_products = product_analysis.sort_values(
    'Gross_Profit',
    ascending=False
)

top_profit_products.head(10)
```

```python
#Low-profit products
low_margin_products = product_analysis.sort_values(
    'Profit Margin %',
    ascending=True
)

low_margin_products.head(10)
```

```python
#Time Analysis
year_analysis = df.groupby('Year').agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).sort_index()

year_analysis
```

```python
#year-over-year growth (%)
year_analysis['Sales Growth %'] = year_analysis['Sales'].pct_change() * 100
year_analysis['Profit Growth %'] = year_analysis['Gross_Profit'].pct_change() * 100

year_analysis
```

```python
year_analysis['Sales Growth %'] = year_analysis['Sales Growth %'].round(2)
year_analysis['Profit Growth %'] = year_analysis['Profit Growth %'].round(2)

year_analysis
```

```python
#Monthly Sales & Profit Trend

#Group by Year and Month
monthly_analysis = df.groupby(
    ['Year', 'Month Number']
).agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).reset_index()

monthly_analysis
```

```python
#Calculate monthly profit margin
monthly_analysis['Profit Margin %'] = (
    monthly_analysis['Gross_Profit'] /
    monthly_analysis['Sales']
) * 100
```

```python
#Sort chronologically
monthly_analysis = monthly_analysis.sort_values(
    ['Year', 'Month Number']
)
monthly_analysis
```

```python
#highest-sales month
best_sales_month = monthly_analysis.loc[
    monthly_analysis['Sales'].idxmax()
]

best_sales_month
```

```python
#lowest-sales month
worst_sales_month = monthly_analysis.loc[
    monthly_analysis['Sales'].idxmin()
]

worst_sales_month
```

```python
#Top-performing regions over time
region_year_analysis = df.groupby(
    ['Year', 'Region']
).agg(
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Gross_Profit=('Gross Profit', 'sum'),
    Units=('Units', 'sum')
).reset_index()

region_year_analysis
```

```python
#Which region grew the most from 2024 to 2025?

#Calculate regional sales growth
region_growth = region_year_analysis.pivot(
    index='Region',
    columns='Year',
    values='Sales'
)

region_growth['Sales Growth %'] = (
    (region_growth[2025] - region_growth[2024])
    / region_growth[2024]
) * 100

region_growth.sort_values('Sales Growth %', ascending=False)
```

```python
#Let's check whether profit grew at the same rate as sales for each region.
region_profit_growth = region_year_analysis.pivot(
    index='Region',
    columns='Year',
    values='Gross_Profit'
)

region_profit_growth['Profit Growth %'] = (
    (region_profit_growth[2025] - region_profit_growth[2024])
    / region_profit_growth[2024]
) * 100

region_profit_growth.sort_values('Profit Growth %', ascending=False)
```
